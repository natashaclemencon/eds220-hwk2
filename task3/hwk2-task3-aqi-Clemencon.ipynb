In this task you will use [Air Quality Index (AQI)](https://www.airnow.gov/aqi/aqi-basics/) data from the [US Environmental Protection Agency](https://www.epa.gov) to visualize the impact on the AQI of the 2017 [Thomas Fire](https://en.wikipedia.org/wiki/Thomas_Fire) in Santa Barbara County.

```python
import pandas as pd
# Read in both datasets. 
aqi_17 = pd.read_csv('daily_aqi_by_county_2017.zip', compression='zip')
aqi_18 = pd.read_csv('daily_aqi_by_county_2018.zip', compression='zip')
# Concatenate the aqi_17 and aqi_18 dataframes to get one df for both 2017 and 2018.
aqi = pd.concat([aqi_17, aqi_18])
# Simplify column names.
aqi.columns = (aqi.columns.str.lower().str.replace(' ','_'))
# Create new aqi_sb DF by filtering aqi for where county_name is Santa Barbara,
# then drop columns state_name, county_name, state_code, and county_code. 
aqi_sb = aqi[aqi['county_name']=='Santa Barbara'].drop(columns=['state_name','county_name', 'state_code', 'county_code'])
# Change the date column in aqi_sb into a datetime64 type.
aqi_sb['date'] = pd.to_datetime(aqi_sb['date'])
# Set the date column of aqi_sb as the index. 
aqi_sb = aqi_sb.set_index('date').sort_index()
# Add a new column that calculates the 5 day rolling average of the Santa Barbara aqi.
aqi_sb['five_day_average'] = aqi_sb['aqi'].rolling('5D').mean()
# Plot the aqi and five day average of aqi_sb. 
# Notice a clear increase in AQI during the Thomas Fire.
aqi_sb.plot(kind='line', y=['aqi', 'five_day_average'], xlabel='Date', ylabel='AQI')
```

In the next cell we will calculate an average over a [rolling window](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.Series.rolling.html) using the `rolling()` method for `pandas.Series`:

- `rolling()` is a lazy method, so we need to specify what we want to calculate over each window before it does something. 
- in this example we use the aggregator function `mean()` to calculate the average over each window
- the parameter '5D' indicates we want the window for our rolling average to be 5 days. 
- we get a `pandas.Series` as output

Store your answer in the `rolling_average` variable. Your result should be a `pandas.Series` indexed by `date`, with the 5-day averages as its values.
