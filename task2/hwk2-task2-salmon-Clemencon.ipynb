## About the data

In this task you will use simplified data from the Alaska Department of Fish & Game containing commercial salmon catch data from 1878 to 1997. The original data can be accessed from the KNB repository:

> [Mike Byerly. (2016). Alaska commercial salmon catches by management region (1886-1997).](https://knb.ecoinformatics.org/view/df35b.304.2) Gulf of Alaska Data Portal. df35b.304.2.

The simplified dataset is in CSV format in the homework repository and has the following columns:

| Column | Description |
| ------ | ----------- | 
| Region | Region code |
| Year | Year fish were caught |
| notesRegCode | Notes and comments |
| Species | Species of salmon caught |
| Catch | Commercial catches of salmon species (in thousands of fish) |

```python
# FINAL CODE
# Import pandas, numpy, and matplotlib.
import pandas as pd
import numpy as np
import matplotlib as plt
# Read salmon data into catch_data.
catch_data = pd.read_csv('data/salmon_data.csv')
# Get rid of the notesRegCode column.
catch_data = catch_data.drop(columns=['notesRegCode'])
# Update the 'Catch' column in row 400 from 'i' to string '1'.
catch_data.at[400, 'Catch'] = '1'
# Update the catch column to be integers. 
catch_data['Catch']=catch_data['Catch'].astype('int64')
# Create a df of average salmon catch counts by region using the groupby function.
avg_region = catch_data.groupby('Region')['Catch'].mean()
# Make avg_region a dataframe, sort it in ascending order, and plot it as a bar graph. 
avg_region_df = pd.DataFrame(avg_region).sort_values(ascending=True, by='Catch').plot(kind='bar',
                            title='Salmon Catches by Region', ylabel='Catch (thousands of salmon')
```
